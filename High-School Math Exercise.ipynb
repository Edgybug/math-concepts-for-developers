````python
# Optional: If you want, you can write your own Python function which calculates the roots.

def find_discriminant(a, b, c):
    return b**2 - 4*a*c 

def solve_quadratic_equation(a, b, c):

    #check if equation is linear
    if a == 0:
        return [-c / b] if b != 0 else []
    
    discriminant = find_discriminant(a, b, c)

    #check if discriminant is < 0
    if discriminant < 0:
        return []

    #check if discriminant is 0
    if discriminant == 0:
        return [-b / (2*a)]

    #calculate both roots
    root1 = (-b + sqrt(discriminant)) / (2 * a)
    root2 = (-b - sqrt(discriminant)) / (2 * a)

    return [root1, root2]

print(solve_quadratic_equation(5, 8, 3))
````

### Problem 5. Intro to proofs: proof by induction
Look up what a _proof by mathematical induction_ is. I like [Brilliant](https://brilliant.org/wiki/writing-a-proof-by-induction/), but you can use whatever you like.

Now, look at this proposition (which we also call _theorem_).
We call **triangular numbers** numbers which represent objects arranged in a triangle, like this [example](https://byjus.com/maths/triangular-numbers/). We denote the $n$ th triangular number by $T_n$.

We can see that one way to write $T_n$ is by summing the numbers from $1$ to $n$. Another way is:
$$ T_n = \frac{n(n+1)}{2}$$

Prove this by induction, starting with $n = 1$.

````python

sp.init_printing()

# set the needed symbols
n = sp.symbols('n', integer=True, positive=True)

# define the formula we will use induction on
formula = n * (n + 1) / 2

# solve for n = 1 - base case
left_side = 1
right_side = formula.subs(n, 1)

# print results
print(f"=== Base Case (n = 1) ===")
print("\nFormula:")
sp.pprint(formula)
print(f"\nLeft side = {left_side}")
print(f"Right side = {right_side}")
print(f"Is it true? -> {left_side == right_side}\n")

# Solve for n = n + 1
next_n = n + 1
induction_left = formula + next_n
induction_right = formula.subs(n, next_n)

# We find the difference between left and right, if 0, both sides are equal
difference = sp.simplify(induction_left - induction_right)

# print results 
print("=== Induction Step (n -> n + 1) ===")
print("\nInduction Left:")
sp.pprint(induction_left)

print("\nInduction Right:")
sp.pprint(induction_right)

print(f"\nDifference between left and right: {difference}")
if difference == 0:
    print("✓ Success! The formula holds true for all n.")
else:
    print("✗ Failed. The induction step does not match.")
````

### * Problem 6. Similarity and sets
**Sets** are collections of unique objects. We can "compare" two sets (or see how similar they are) using the metric _intersection over union_.

You're on your own now! Try the following operations, following the guidelines below.
* What is a set? How do we write sets in Python?
    - Set is a collection of unique values
    - In python set is written set = {value1, value2, value3}
* What is set union and set intersection?
    - Union is combining 2 or more sets. if there are repeating values in different sets, only 1 is kept as they are unique
    - Intersection - shows the common values between 2 or more sets
* How is IoU defined? What does it measure?
    - IoU is defined by dividing the union with the intersection, 
    typically a number between 0 and 1 - 
    shows the percentage of overlap between the sets
* Example 1. Create two sets of names and compute their IoU.
* Example 2. Consider two intervals on the number line, e.g., $[2;\ 3]$ and $[1; 2.5]$. What is their IoU?
* Example 3. Consider two rectangles within an image (parallel to the image axes). Each rectangle contains a certain amount of pixels. How do you describe and compute their IoU? We use this daily in tasks like **object detection**. [Hint](https://pyimagesearch.com/2016/11/07/intersection-over-union-iou-for-object-detection/)

````python
# Example 1 - Names 

sci_fi_characters = {"Luke", "Spock", "Ellen", "Neo", "Aragorn", "Arthur"}
fantasy_characters = {"Frodo", "Gandalf", "Harry", "Percy", "Luke", "Spock", "Ellen"}

intersection = sci_fi_characters & fantasy_characters
union = sci_fi_characters |  fantasy_characters

# IoU is calculated  based on the lenght of the sets
IoU = len(intersection) / len(union)

print("Intersection:", intersection)
print("Union:", union)
print(f"IoU: { IoU * 100}%") 
````

````python
# Example 2. Consider two intervals on the number 
# line, e.g., [2; 3] and [1; 2.5] What is their IoU?

A = (2,3)
B = (1,2.5)

#find the endpoints of the intersection
left_endpoint = max(A[0], B[0])
right_endpoint = min(A[1], B[1])

#calculate intersection length
intersection = right_endpoint - left_endpoint

#calculate the union len (the whole length from 1 to 3 is 2)
union = max(A) - min(B)

IoU = intersection / union
print(f"IoU: {IoU}")

# Draw the lines to visualise the intersection
plt.figure(figsize=(3, 2))
plt.hlines(y=1, xmin=A[0], xmax=A[1], color='r')
plt.hlines(y=1.2, xmin=B[0], xmax=B[1], color='b')
plt.vlines(A[0], 0, 2, linestyles ="solid", colors ="g")
plt.vlines(B[1], 0, 2, linestyles ="solid", colors ="g")
plt.show()
````
